# D11 实操区 · 模块 8 范式判断 1NF~BCNF（纯理论，纸面作答）

📌 **本模块没有 SQL**。③应用题把**判断 + 理由**写在 markdown 格子里。

**判断六步**：找候选键 → 分主/非主属性 → 1NF → 2NF → 3NF → BCNF

## ③ 应用题（写出判断 + 理由）

**应用 1**：选课表 `SC(学号, 课程号, 成绩, 姓名)`。已知 `(学号,课程号)→成绩`、`学号→姓名`。问：满足 2NF 吗？为什么？如不满足，怎么拆？

**应用 2**：`R(学号, 系名, 系主任)`。已知 `学号→系名`、`系名→系主任`。问：满足 3NF 吗？为什么？如不满足，怎么拆？

**应用 3**：`R(学生, 课程, 教师)`。已知 `(学生,课程)→教师`、`教师→课程`（每个教师只教一门课，每门课由多个教师教）。问：是 3NF 吗？是 BCNF 吗？分别说明理由。

**应用 4**：`R(学号, 课程号, 成绩, 姓名, 系名, 系主任)`。依赖：`学号→姓名`、`学号→系名`、`系名→系主任`、`(学号,课程号)→成绩`。问：最高满足第几范式？说明每一级为什么卡住（或通过）。

**应用 5**：把 `SC(学号, 课程号, 成绩, 姓名)` 拆成满足 2NF 的关系模式。写出每张表（表名 + 属性），并说明为什么满足 2NF。

## ③ 应用答案（应用 1 ~ 应用 5）

应用1.不满足。2NF的要求是没有部分依赖，由条件可知(学号,课程号)的一部分“学号”能确定姓名，那(学号,课程号)作为候选键也能确定姓名，姓名对(学号,课程号)就是部分依赖，只依赖学号。应该拆成两个表(学号, 姓名)和(学号,课程号, 成绩)。
应用2.不满足。3NF的要求是没有传递依赖，学号确定系名，系名确定系主任，由此得出学号确定系主任，这就是传递依赖。应该拆成(学号, 系名)和(系名, 系主任)两个表。
应用3.是3NF，因为没有非主属性。不是BCNF，因为这要求每一个决定因素都是候选键，教师确定课程，这里教师是决定因素，但是不是候选键。
应用4.满足1NF，因为所有属性原子不可拆分。不满足2NF，因为候选键是(学号, 课程号)，但是单独的学号就可确定姓名、系名，不满足部分依赖。不满足3NF，因为学号确定系名，系名确定系主任，所以这里有了传递依赖。不满足BCNF，因为没有满足3NF。
应用5.学生表(学号，姓名)，成绩表(学号，课程号，成绩)。学生表的候选键是学号，单属性，天然满足2NF，不可能出现部分依赖，成绩表的候选键是(学号，课程号)，单独学号不能确定成绩，单独课程号也不能确定成绩，所以没有部分依赖，也是2NF。
