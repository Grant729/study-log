````python
%%sql
SELECT 
  COUNT(*) AS 读者总数,
  SUM(借阅次数) AS 借阅次数总和,
  AVG(借阅次数) AS 平均借阅次数,
  MAX(借阅次数) AS 最高,
  MIN(借阅次数) AS 最低
FROM 读者;
````
