```python
%%sql
CREATE TABLE 借阅_模拟 (
  借阅号 INT PRIMARY KEY AUTO_INCREMENT,
  读者编号 CHAR(3),
  图书编号 CHAR(4),
  借阅日期 DATE,
  应还日期 DATE NOT NULL,
  FOREIGN KEY (读者编号) REFERENCES 读者(读者编号),
  FOREIGN KEY (图书编号) REFERENCES 图书(图书编号)
);
DESCRIBE 借阅_模拟;
```

```python
%%sql
DROP TABLE IF EXISTS 借阅_模拟;
SHOW TABLES LIKE '借阅_模拟';
```

```python
%%sql
SELECT 读者姓名, 借阅次数
FROM 读者
WHERE 借阅次数 = (
  SELECT MAX(借阅次数)
  FROM 读者
);
```

```python
%%sql
SELECT
  读者.读者姓名 AS 姓名,
  COUNT(借阅.图书编号) AS 借阅本数
FROM 读者
LEFT JOIN 借阅
ON 读者.读者编号 = 借阅.读者编号
GROUP BY 读者.读者编号, 读者.读者姓名
ORDER BY 借阅本数 DESC;
```

```python
%%sql
DROP PROCEDURE IF EXISTS p_借阅本数;
CREATE PROCEDURE p_借阅本数(
  IN p_编号 CHAR(3),
  OUT p_本数 INT
)
BEGIN
  SELECT COUNT(图书编号)
  INTO p_本数
  FROM 借阅
  WHERE 读者编号 = p_编号;
END;
CALL p_借阅本数('001', @本数);
SELECT @本数 AS 借阅本数;
```

```python
%%sql
DROP TRIGGER IF EXISTS trg_删借阅减次数;
CREATE TIRGGER trg_删借阅减次数
AFTER DELETE ON 借阅
FOR EACH ROW
BEGIN
  UPDATE 读者
  SET 借阅次数 = 借阅次数 - 1
  WHERE 读者编号 = OLD.读者编号;
END;
```
