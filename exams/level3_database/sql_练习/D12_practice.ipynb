## ③ 应用题（写出判断 + 理由 + 验证）

**应用 1**：`R(学号, 姓名, 班级)`，FD：`学号→姓名`、`学号→班级`。分解为 `R1(学号,姓名)` + `R2(姓名,班级)`。问：无损还是有损？用定理判断，并说明后果。

**应用 2**：`R(学号, 系名, 系主任)`，FD：`学号→系名`、`系名→系主任`。分解为 `R1(学号,系名)` + `R2(学号,系主任)`。问：① 是否无损连接？② 是否保持函数依赖？给出结论。

**应用 3**：承应用 2 的 `R(学号,系名,系主任)`。要求：拆成**既无损、又保持依赖**的 3NF，并**逐条验证**两个条件。

**应用 4**：`R(学号, 课程号, 成绩, 姓名)`，FD：`(学号,课程号)→成绩`、`学号→姓名`。要求：拆成 2NF，并验证无损连接 + 保持函数依赖。

**应用 5**：`R(借阅号, 读者编号, 读者姓名, 图书编号, 图书名)`，FD：`借阅号→读者编号`、`读者编号→读者姓名`、`借阅号→图书编号`、`图书编号→图书名`。
问：① 候选键？非主属性？② 最高第几范式？理由？③ 拆成 3NF（无损 + 保持依赖）。

## ③ 应用答案（应用 1 ~ 应用 5）

应用1.有损，分解出来两个表的交集属性是姓名，差是学号或班级，姓名不能确定学号，可能有重名，同理也不能确定班级，所以有损。
应用2.无损，交集属性是学号，差属性是系名和系主任，学号可以确定系名，学号也可以确定系主任。不保持函数依赖，原函数依赖学号确定系名可以在R1中找到，但是系名确定系主任却已经没有啦。
应用3.(学号，系名)，(系名，系主任)。公共属性是系名，差属性是系主任或学号，系名可以确定系主任，是无损。原两个依赖学号确定系名和系名确定系主任分别在两个新表里，保持函数依赖。
应用4.(学号，姓名)，(学号，课程号，成绩)。公共属性是学号，R1-R2是姓名，学号确定姓名，无损连接。原函数依赖分别在两个新表里，保持函数依赖。
应用5.候选键是借阅号，非主属性有读者编号、图书编号、读者姓名、图书名。满足2NF，候选键都是单属性，天然没有部分依赖，借阅号确定读者编号确定读者姓名，借阅号确定图书编号确定图书名，这里有两个传递依赖，不满足3NF。拆成3NF分别是R1(借阅号，读者编号，图书编号)，R2(读者编号，读者姓名)，R3(图书编号，图书名)，R1、R2公共属性是读者编号，R2-R1是读者姓名，读者编号确定读者姓名，满足无损，R1和R3同理，原四个依赖如果编号1234，依赖1和3在R1，依赖2在R2，依赖4在R3，满足保持依赖。
