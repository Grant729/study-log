## ③ 应用题（判断 + 一句话理由）

**应用 1**：选课表 SC(学号, 课程号, 成绩)。已知 `(学号,课程号) → 成绩`。问：这是完全依赖还是部分依赖？为什么？

**应用 2**：选课表 SC(学号, 课程号, 成绩, 姓名)。已知 `(学号,课程号) → 姓名`，且 `学号 → 姓名`。问：`(学号,课程号) → 姓名` 是完全还是部分依赖？为什么？

**应用 3**：借阅表(借阅号, 读者编号, 读者姓名)。已知 `借阅号 → 读者编号`、`读者编号 → 读者姓名`。问：`借阅号 → 读者姓名` 是什么依赖？按三要素验证。

**应用 4**：选课表 SC(学号, 课程号, 成绩)。问：候选键是什么？为什么是它、而不是学号或课程号单独当键？

## ③ 应用答案（应用 1 ~ 应用 5）

应用1.完全依赖，单单知道学号不能唯一确定成绩，单单知道课程号也不能唯一确定成绩，只有两个同时确定，才是唯一确定成绩
应用2.部分依赖，学号是(学号，课程号)的真子集，已知学号确定姓名，也就是存在真子集确定姓名，所以是部分依赖，在这里课程号是不必要的
应用3.传递依赖，借阅号确定读者编号，读者编号确定读者姓名，读者编号作为一个中间值起到了桥梁的传递作用，同时读者编号不能反向确定唯一借阅号，因为一个读者可以借多本书，也就可以有多个借阅号，所以借阅号和读者编号不等价，传递成立
应用4.候选键是(学号，课程号)。因为学号和课程号加起来能确定唯一成绩，且单独的学号可以有多个成绩，单独的课程号可以有多个人的成绩，即不存在真子集能确定唯一值，满足候选键条件
应用5.候选键是(学号，课程号)，因为二者加起来能确定唯一成绩，且学号、课程号都不能单独确定成绩。系主任传递依赖于学号，因为学号确定系名，系名确定系主任，从而推导出学号确定系主任，学号是起点，系主任是终点，终点依赖于起点。姓名相对于(学号，课程号)是部分依赖，因为单独的学号就能确定姓名，课程号在这里是不必要条件，所以存在(学号，课程号)的真子集学号能确定唯一姓名，此为部分依赖。
