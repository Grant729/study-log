# D13 实操区 · 模块 9 ER 模型 + ER→关系模式转换（纯理论，纸面作答）

📌 **本模块没有 SQL**。③应用题把 **ER 要素 + 关系模式** 写在 markdown 格子里。

**转换四规则**：① 一实体一张表 ② 1:N 不建表、1 的键放进 N ③ M:N 必建表、两键拼成键 ④ 1:1 可合可拆

## ③ 应用题（写出判断 + 理由 + 关系模式）

**应用 1**：判断下列三组场景各是什么联系类型（1:1 / 1:N / M:N），并说明理由：

- a. 班级 — 学生：一个班级有多名学生，一名学生只属于一个班级
- b. 医生 — 病人（诊治）：一名医生可以诊治多名病人，一名病人可以被多名医生诊治
- c. 科室 — 主任：一个科室只有一名主任，一名主任只负责一个科室

**应用 2**：实体：班级(班级号, 班级名)；学生(学号, 姓名)。联系：属于（班级 1 : N 学生）。写出转换后的关系模式，并说明为什么这样放。

**应用 3**：实体：医生(医生编号, 姓名)；病人(病历号, 姓名)。联系：诊治（医生 M : N 病人），联系属性：诊治日期、诊断结果。写出转换后的关系模式，标出主键。

**应用 4**：实体：科室(科室编号, 科室名)；医生(医生编号, 姓名)。联系：负责（科室 1 : 1 医生，即"科室主任"）。给出**两种**转换方案，并说明各自特点。

**应用 5**：场景（学校）——实体：学生(学号, 姓名)、课程(课程号, 课程名)、教师(教师编号, 姓名)、班级(班级号, 班级名)。联系：属于（学生 N—1 班级）、选修（学生 M—N 课程，属性：成绩）、授课（教师 M—N 课程，无联系属性）。
问：① 写出最终关系模式（每张表的列 + 主键）② 一共几张表？哪几个是"联系"转换来的？

## ③ 应用答案（应用 1 ~ 应用 5）

应用1.a是班级1:N学生，因为一个班级有多名学生，一名学生只属于一个班级，b是医生M:N病人，因为一名医生可以诊治多名病人，一名病人可以被多名医生诊治，c是科室1:1主任，因为一个科室只有一名主任，一名主任只负责一个科室
应用2.(班级号，班级名)(学号，姓名，班级号)，两个实体分别建一张表，因为关系是班级 1 : N 学生，作为1的班级将自己的主键作为外键放到作为N的学生表里
应用3.医生表(医生编号*主键，姓名)，病人表(病历号*主键，姓名)，诊治表(医生编号*主键，病历号*主键，诊治日期，诊断结果)，医生和病人两个实体单独建表，M:N关系单独建表，以两实体的键为主键，也是外键
应用4.一种是合并，科室表(科室编号，科室名，主任医生编号)，医生表(医生编号，姓名)，少一张表，查询方便。一种是单独建表，科室表(科室编号，科室名)，医生表(医生编号，姓名)，负责表(科室编号，主任医生编号)，多一张表，后面添加列方便修改维护
应用5.下面用*代表主键，学生表(学号*，姓名，班级号)，课程表(课程号*，课程名)，教师表(教师编号*，姓名)，班级表(班级号*，班级名)，成绩表(学号*，课程号*，成绩)，授课表(教师编号*，课程号\*)，一共有6张表，学生表里的班级号属性是联系1“属于”转换来的，成绩表和授课表分别是联系2和3转换来的
