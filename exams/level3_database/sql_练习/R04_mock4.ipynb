## 三、题 42（SQL · 超市销售）

（1）建表：

（3）查询 A（销售额）+ 查询 B（库存 NULL 或 0）：


```python
%%sql
CREATE TABLE 商品(
  商品号 CHAR(3) PRIMARY KEY, 
  商品名 VARCHAR(20) NOT NULL, 
  单价 DECIMAL(4, 2), 
  库存 INT
);
CREATE TABLE 销售明细(
  销售单号 CHAR(3), 
  商品号 CHAR(3), 
  数量 INT, 
  销售日期 DATE,
  PRIMARY KEY (销售单号, 商品号),
  FOREIGN KEY (商品号) REFERENCES 商品(商品号)
);
SHOW TABLES;
```

```python
%%sql
INSERT INTO 商品 VALUES
  ('P01', '洗衣液', 39.90, 50),
  ('P02', '牙膏', 12.50, NULL),
  ('P03', '纸巾', 5.00, 0),
  ('P04', '洗发水', 45.00, 20);
SELECT * FROM 商品;
```

```python
%%sql
INSERT INTO 销售明细 VALUES
  ('S01', 'P01', 2, '2026-10-24'),
  ('S01', 'P03', 10, '2026-10-24'),
  ('S02', 'P04', 1, '2026-10-24');
SELECT * FROM 销售明细;
```

```python
%%sql
SELECT 
  商品.商品名 AS 商品名,
  SUM(单价 * 数量) AS 销售额
FROM 商品 LEFT JOIN 销售明细
ON 商品.商品号 = 销售明细.商品号
GROUP BY 商品.商品号, 商品.商品名
ORDER BY 销售额 DESC;
```

```python
%%sql
SELECT 商品号, 商品名, 库存
FROM 商品
WHERE 库存 = 0 
OR 库存 IS NULL;
```

## 三、题 43（存储过程 + 触发器 · 校园一卡通）

（1）建表：

（2）插数据（历史记录）：

（3）触发器：

（4）存储过程 + 调用：

（5）测试结果：


```python
%%sql
CREATE TABLE 学生(
  学号 CHAR(7) PRIMARY KEY, 
  姓名 VARCHAR(10) NOT NULL, 
  余额 DECIMAL(4, 2)
);
CREATE TABLE 消费记录(
  记录号 INT PRIMARY KEY AUTO_INCREMENT,
  学号 CHAR(7),
  金额 DECIMAL(4, 2),
  消费时间 DATETIME,
  FOREIGN KEY (学号) REFERENCES 学生(学号)
);
SHOW TABLES;
```

```python
%%sql
INSERT INTO 学生 VALUES
  ('2023001', '张三', 79.50),
  ('2023002', '李四', 30.00);
SELECT * FROM 学生;
```

```python
%%sql
INSERT INTO 消费记录 VALUES
  (0, '2023001', 12.50, '2026-10-20 12:00:00'),
  (0, '2023001', 8.00, '2026-10-21 08:00:00'),
  (0, '2023002', 20.00, '2026-10-21 18:00:00');
SELECT * FROM 消费记录;
```

```python
%%sql
CREATE TRIGGER trg_消费扣款
AFTER INSERT ON 消费记录
FOR EACH ROW
BEGIN
  UPDATE 学生
  SET 余额 = 余额 - NEW.金额
  WHERE 学号 = NEW.学号;
END;
```

```python
%%sql
CREATE PROCEDURE p_消费总额(
  IN p_学号 CHAR(7),
  OUT p_消费总金额 DECIMAL(4, 2)
)
BEGIN
  SELECT SUM(金额)
  INTO p_消费总金额
  FROM 消费记录
  WHERE 学号 = p_学号;
END;
```

```python
%%sql
INSERT INTO 消费记录 VALUES
  (0, '2023001', 10.00, '2026-10-25 09:00:00');
SELECT 余额 
FROM 学生
WHERE 学号 = '2023001';
```

```python
%%sql
SET @学号 = '2023001';
CALL p_消费总额(@学号, @消费总金额);
SELECT @学号 AS 学号, @消费总金额 AS 消费总金额;
```
