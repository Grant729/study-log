```python
%%sql
DROP USER IF EXISTS 'd17_stu'@'localhost';
CREATE USER 'd17_stu'@'localhost'
IDENTIFIED BY 'Stu@12345';
SHOW GRANTS FOR 'd17_stu'@'localhost';
```

```python
%%sql
GRANT SELECT
ON exam_practice.图书
TO 'd17_stu'@'localhost';
SHOW GRANTS FOR 'd17_stu'@'localhost';
```

```python
%%sql
GRANT SELECT, INSERT
ON exam_practice.借阅
TO 'd17_stu'@'localhost';
SHOW GRANTS FOR 'd17_stu'@'localhost';
```

```python
%%sql
REVOKE INSERT
ON exam_practice.借阅
FROM 'd17_stu'@'localhost';
SHOW GRANTS FOR 'd17_stu'@'localhost';
```
