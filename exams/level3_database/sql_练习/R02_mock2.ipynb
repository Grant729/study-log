```python
%%sql
DROP TABLE IF EXISTS 停车记录;
DROP TABLE IF EXISTS 车位;
CREATE TABLE 车位(
  车位号 CHAR(3) PRIMARY KEY,
  区域 VARCHAR(10) NOT NULL,
  类型 VARCHAR(10) NOT NULL
);
CREATE TABLE 停车记录(
  记录号 INT PRIMARY KEY AUTO_INCREMENT,
  车位号 CHAR(3) NOT NULL,
  车牌号 VARCHAR(10) NOT NULL,
  入场时间 DATETIME NOT NULL,
  出场时间 DATETIME,
  FOREIGN KEY (车位号) REFERENCES 车位(车位号)
);
SHOW TABLES FROM exam_practice;
```

```python
%%sql
SELECT * FROM 停车记录;
```

```python
%%sql
SELECT 车牌号, 车位号, 入场时间
FROM 停车记录
WHERE 出场时间 IS NULL
ORDER BY 入场时间;
```

```python
%%sql
SELECT 区域, COUNT(车位号) AS 车位数
FROM 车位
GROUP BY 区域
ORDER BY 车位数 DESC;
```

```python
%%sql
DROP TABLE IF EXISTS 报名;
DROP TABLE IF EXISTS 学员;
CREATE TABLE 学员(
  学员号 CHAR(3) PRIMARY KEY,
  姓名 VARCHAR(20) NOT NULL,
  报名数 INT
);
CREATE TABLE 报名(
  学员号 CHAR(3),
  课程号 CHAR(4),
  PRIMARY KEY (学员号, 课程号),
  FOREIGN KEY (学员号) REFERENCES 学员(学员号)
);
SHOW TABLES;
```

```python
%%sql
INSERT INTO 学员 VALUES
  ('S01', '小明', 0),
  ('S02', '小红', 0);
SELECT * FROM 学员;
```

```python
%%sql
DROP PROCEDURE IF EXISTS p_报名数;
CREATE PROCEDURE p_报名数 (
  IN p_学员号 CHAR(3),
  OUT p_报名课程数 INT
)
BEGIN
  SELECT 报名数
  INTO p_报名课程数
  FROM 学员
  WHERE 学员号 = p_学员号;
END;
SET @学员号 = 'S01';
CALL p_报名数(@学员号, @报名数);
SELECT @学员号 AS 学员号, @报名数 AS 报名数;
```

```python
%%sql
DROP TRIGGER IF EXISTS trg_报名加一;
CREATE TRIGGER trg_报名加一
AFTER INSERT ON 报名
FOR EACH ROW
BEGIN
  UPDATE 学员
  SET 报名数 = 报名数 + 1
  WHERE 学员号 = NEW.学员号;
END;
```
