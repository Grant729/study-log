```python
%%sql
DROP TABLE IF EXISTS 点单;
DROP TABLE IF EXISTS 菜品;
CREATE TABLE 菜品 (
  菜品号 CHAR(3) PRIMARY KEY,
  菜名 VARCHAR(20) NOT NULL,
  单价 DECIMAL(4, 2) NOT NULL,
  分类 VARCHAR(10) NOT NULL
);
CREATE TABLE 点单 (
  订单号 CHAR(4),
  菜品号 CHAR(3),
  数量 INT NOT NULL,
  下单时间 DATETIME NOT NULL,
  PRIMARY KEY (订单号, 菜品号),
  FOREIGN KEY (菜品号) REFERENCES 菜品(菜品号)
);
SHOW TABLES;
```

```python
%%sql
DELETE FROM 点单;
DELETE FROM 菜品;
INSERT INTO 菜品 VALUES
  ('D01', '宫保鸡丁', 38.00, '热菜'),
  ('D02', '麻婆豆腐', 28.00, '热菜'),
  ('D03', '凉拌黄瓜', 12.00, '凉菜'),
  ('D04', '米饭', 3.00, '主食');
INSERT INTO 点单 VALUES
  ('O001', 'D01', 2, '2026-10-22 12:00:00'),
  ('O001', 'D02', 1, '2026-10-22 12:00:00'),
  ('O002', 'D01', 1, '2026-10-22 18:30:00'),
  ('O003', 'D04', 3, '2026-10-22 18:30:00');
SELECT * FROM 菜品;
```

```python
%%sql
SELECT * FROM 点单;
```

```python
%%sql
SELECT 菜名, SUM(单价 * 数量) AS 销售额
FROM 菜品 RIGHT JOIN 点单
ON 菜品.菜品号 = 点单.菜品号
GROUP BY 菜品.菜品号, 菜品.菜名
ORDER BY 销售额 DESC;
```

```python
%%sql
SELECT 菜名 FROM 菜品
WHERE 菜品号 NOT IN (
  SELECT 菜品号 FROM 点单
);
```

```python
%%sql
DROP TABLE IF EXISTS 观看记录;
DROP TABLE IF EXISTS 用户;
CREATE TABLE 用户(
  用户号 CHAR(3) PRIMARY KEY,
  昵称 VARCHAR(10) NOT NULL,
  观看次数 INT
);
CREATE TABLE 观看记录(
  记录号 INT PRIMARY KEY AUTO_INCREMENT,
  用户号 CHAR(3),
  影片号 CHAR(4),
  观看日期 DATE,
  FOREIGN KEY (用户号) REFERENCES 用户(用户号)
);
SHOW TABLES; 
```

```python
%%sql
INSERT INTO 用户 VALUES
  ('U01', '小明', 2),
  ('U02', '小红', 1);
INSERT INTO 观看记录 VALUES
  (0, 'U01', 'M001', '2026-10-20'),
  (0, 'U01', 'M002', '2026-10-21'),
  (0, 'U02', 'M001', '2026-10-21');
SELECT * FROM 用户;
```

```python
%%sql
SELECT * FROM 观看记录;
```

```python
%%sql
CREATE TRIGGER trg_删观看减次数
AFTER DELETE ON 观看记录
FOR EACH ROW
BEGIN
  UPDATE 用户
  SET 观看次数 = 观看次数 - 1
  WHERE 用户号 = OLD.用户号;
END;
```

```python
%%sql
CREATE PROCEDURE p_观看数(
  IN p_用户号 CHAR(3),
  OUT p_观看记录数 INT
)
BEGIN
  SELECT COUNT(*) 
  INTO p_观看记录数
  FROM 观看记录
  WHERE 用户号 = p_用户号;
END;
```

```python
%%sql
CALL p_观看数('U01', @观看数);
SELECT @观看数 AS 观看数;
```

```python
%%sql
DELETE FROM 观看记录
WHERE 影片号 = 'M002';
SELECT 观看次数
FROM 用户
WHERE 用户号 = 'U01';
```
