## ③ 概念应用答案（应用 1 ~ 应用 5）

应用1.a对应volume，代表数据的大量性，b对应velocity，代表数据的快速性，c对应variety，代表数据的多样性，d对应value，代表数据的价值性。
应用2.a是结构化，一张二维表是结构化，b是半结构化，json数据是典型代表，c是非结构化，视频、音频、图片、文字都是非结构化，d是半结构化，有一定的结构。
应用3.MapReduce是批处理，storm、flink是流计算，Pregel、giragh是图计算，hive是查询分析计算。
应用4.a是DataNode，b是NameNode，c是secondarynamenode，d是hive，e是sqoop。
应用5.①是批处理，海量数据+离线统计，②是流计算，要求做到数据持续+低延迟，③是图计算，社交关系网络是一张巨大的图，有节点和边。
