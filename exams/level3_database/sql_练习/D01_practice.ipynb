## 题 10：条件查询 + 排序

再新建一个代码格（`%%sql` 开头），查出借阅量 ≥ 5 的读者，按借阅量从高到低排。

> 提示：`%%sql` 双百分号开头的格子，内容整体当 SQL 执行；报错就截图发教练。

```python
%%sql
CREATE DATABASE exam_practice;
USE exam_practice;
CREATE TABLE 读者 (
    读者编号 CHAR(3) PRIMARY KEY,
    读者姓名 VARCHAR(20) NOT NULL,
    借阅次数 INT DEFAULT 0
);
INSERT INTO 读者 VALUES
    ('001', '张三', 5),
    ('002', '李四', 8),
    ('003', '王五', 3),
    ('004', '赵六', 2);
```

```python
%%sql
SELECT * FROM 读者 WHERE 借阅次数 >= 5 ORDER BY 借阅次数 DESC;
```
