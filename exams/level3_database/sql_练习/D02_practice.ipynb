```python
%%sql
DROP TABLE IF EXISTS 读者_大一;
CREATE TABLE 读者_大一 (
    编号 CHAR(3) PRIMARY KEY,
    姓名 VARCHAR(20)
);
INSERT INTO 读者_大一 VALUES
    ('001','张三'),
    ('002','李四'),
    ('003','王五');
DROP TABLE IF EXISTS 读者_大二;
CREATE TABLE 读者_大二 (
    编号 CHAR(3) PRIMARY KEY,
    姓名 VARCHAR(20)
);
INSERT INTO 读者_大二 VALUES
    ('101','李四'),
    ('102','赵六'),
    ('103','钱七');
SELECT * FROM 读者_大一 UNION SELECT * FROM 读者_大二;
```

```python
%%sql
SELECT 读者姓名 FROM 读者 WHERE 借阅次数 > 3;
```
