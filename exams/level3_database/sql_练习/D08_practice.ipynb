````python
%%sql
SELECT 书名, 馆藏量
FROM 图书
WHERE 馆藏量 = (
    SELECT MAX(馆藏量)
    FROM 图书
);
````

````python
%%sql
SELECT 书名, 馆藏量
FROM 图书
WHERE 馆藏量 > (
    SELECT AVG(馆藏量)
    FROM 图书
);
````

````python
%%sql
SELECT 书名
FROM 图书
WHERE 图书编号 IN (
    SELECT 图书编号
    FROM 借阅
    WHERE 读者编号 IN (
        SELECT 读者编号
        FROM 读者
        WHERE 借阅次数 >= 3
    )
);
````

````python
%%sql
SELECT 书名, 馆藏量
FROM 图书
WHERE 馆藏量 > ALL (
    SELECT 馆藏量
    FROM 图书
    WHERE 出版社 = '人民邮电出版社'
);
````

````python
%%sql
SELECT 书名
FROM 图书
WHERE EXISTS (
    SELECT 1
    FROM 借阅
    WHERE 图书.图书编号 = 借阅.图书编号
);
````

````python
%%sql
SELECT 书名
FROM 图书
WHERE NOT EXISTS (
    SELECT 1
    FROM 借阅
    WHERE 图书.图书编号 = 借阅.图书编号
);
````
