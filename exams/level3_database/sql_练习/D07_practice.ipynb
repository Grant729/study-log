````python
%%sql
SELECT
  读者编号,
  COUNT(*) AS 借阅本数
FROM 借阅
GROUP BY 读者编号
ORDER BY 读者编号;
````

````python
%%sql
SELECT
  读者编号,
  MAX(应还日期) AS 最晚应还
FROM 借阅
GROUP BY 读者编号
ORDER BY 读者编号;
````

````python
%%sql
SELECT
  读者编号,
  COUNT(*) AS 本数
FROM 借阅
WHERE 应还日期 <= '2026-11-20'
GROUP BY 读者编号
ORDER BY 读者编号;
````

````python
%%sql
SELECT
  读者编号,
  COUNT(*) AS 借阅本数
FROM 借阅
GROUP BY 读者编号
HAVING COUNT(*) >= 2;
````

````python
%%sql
SELECT
  读者.读者姓名, 
  COUNT(借阅.借阅号) AS 借阅本数
FROM 读者
LEFT JOIN 借阅
ON 读者.读者编号 = 借阅.读者编号
GROUP BY 读者.读者姓名
ORDER BY 借阅本数 DESC;
````
