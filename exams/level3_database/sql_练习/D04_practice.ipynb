## 题 12：ALTER TABLE 三新招

一个代码格（从上到下）：

```sql
DROP TABLE IF EXISTS 课程;
CREATE TABLE 课程 (课程号 CHAR(3) PRIMARY KEY, 课程名 VARCHAR(30) NOT NULL, 学分 INT);
INSERT INTO 课程 VALUES ('C01','数据库系统概论',4),('C02','信息检索',3),('C03','Python程序设计',3);
ALTER TABLE 课程 ADD COLUMN 授课教师 VARCHAR(20) DEFAULT '待定';
ALTER TABLE 课程 MODIFY 学分 DECIMAL(3,1);
ALTER TABLE 课程 CHANGE 授课教师 教师姓名 VARCHAR(20);
DESCRIBE 课程;
SELECT * FROM 课程;
ALTER TABLE 课程 DROP COLUMN 教师姓名;
DESCRIBE 课程;
```

**验收**：第一个 DESCRIBE 显示 4 列（学分变 decimal(3,1)）；SELECT 里教师姓名全是"待定"；最后 DESCRIBE 回到 3 列。

## 题 13：视图三连

一个代码格：`CREATE VIEW 高借阅读者 AS SELECT * FROM 读者 WHERE 借阅次数 > 3;` → `SELECT * FROM 高借阅读者;` → `UPDATE 高借阅读者 SET 借阅次数 = 10 WHERE 读者编号 = '001';` → `SELECT * FROM 读者 WHERE 读者编号 = '001';`（应看到张三变成 10）→ `UPDATE 读者 SET 借阅次数 = 5 WHERE 读者编号 = '001';`（恢复）→ `DROP VIEW 高借阅读者;`

## 回 6：ALTER + CHECK 换汤
