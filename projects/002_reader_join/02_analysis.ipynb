```r
library(ggplot2)
df <- read.csv("data/merged.csv", fileEncoding = "UTF-8-BOM")
df <- df[!is.na(df$年级), ]

agg <- aggregate(借阅量 ~ 年级, data = df, FUN = sum)
agg <- agg[order(agg$借阅量, decreasing = TRUE), ]
agg

ggplot(agg, aes(x = reorder(年级, -借阅量), y = 借阅量)) +
  geom_col(fill = "steelblue") +
  theme_minimal()

ggsave("outputs/年级借阅量.png", width = 8, height = 5)
```
